```python
import numpy as np

eps = 0.0000000001
inf = 1000000000000000000000000000000
```

```python
n = 7
v = np.array([4, 1, 2, 6, 3, 6, 2], dtype=int)
c = np.array([9, 8, 6, 9, 2, 5, 4], dtype = float)
B = 15
```

```python
opt = np.zeros(shape=(n, B + 1), dtype=float)
x = np.zeros(shape=(n, B + 1), dtype=int)

for k in range(n):
    for b in range(B + 1):
        if k == 0:
            if v[k] <= b:
                opt[k][b] = c[k]
                x[k][b] = 1
            else:
                opt[k][b] = 0
                x[k][b] = 0
        else: # k >= 2
            if v[k] <= b:
                opt1 = opt[k - 1][b]
                opt2 = c[k] + opt[k - 1][b - v[k]]
                if opt1 < opt2:
                    opt[k][b] = opt2
                    x[k][b] = 1
                else:
                    opt[k][b] = opt1
                    x[k][b] = 0
            else:
                opt[k][b] = opt[k - 1][b]
                x[k][b] = 0
```

```python
print("opt = ")
print(opt)

print("x = ")
print(x)
```

```python
brali = np.zeros(shape=(n,))

k = n - 1
b = B

while k >= 0:
    brali[k] = x[k][b]

    if brali[k] == 1:
        b = b - v[k]
    k = k - 1

print(f"brali = {brali}")
```
